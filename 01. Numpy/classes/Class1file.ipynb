# Numpy Arrays

NumPy arrays are the main way we will use Numpy throughout the course. Numpy arrays essentially come in two flavors: vectors and matrices. Vectors are strictly 1-d arrays and matrices are 2-d (but you should note a matrix can still have only one row or one column).

Let's begin our introduction by exploring how to create NumPy arrays.

## Creating NumPy Arrays

### From a Python List

We can create an array by directly converting a list or list of lists:

```python
import numpy as np
```

Lets create a matrix using list

```python
l = [1,2,3,4,5]
arr = np.array(l)
print(arr*5)
```

Now lets see how matlix looks like using numpy arrays

```python
l = [[1,2,3,4],[5,6,7,8]]
arr = np.array(l)
print(arr)
```

## Built-in Methods

there are lots of build in methods to create an array lets create some and see

np.zeros(), np.ones(), np.full()

np.arange(), np.linspace()

np.eye()

```python
arr = np.zeros(4)   #0. - indicates that methods is made using built in function
print(arr)
```

```python
print(np.zeros([4,4]))  #[no. of rows, no. of columns]
```

```python
arr = np.ones(4)
print(arr)
```

```python
arr = np.ones([2,2])
print(arr)
```

```python
arr = np.full(5,7)     #for creating custom array we use .full
print(arr)
```

```python
arr = np.arange(1,11,2)  #same as range
arr
```

```python
arr = np.linspace(0,1,5)    #evenly distributed 5 values
print(arr)
```

```python
arr = np.eye(6)
print(arr)
```

```python
arr = np.random.rand(3,3)    #rand(rows,columns)
arr                          #float
```

```python
arr = np.random.randint(1,10,(3,3))     #(start,stop,(shape))
print(arr)                              #integer
```

```python
arr = np.random.randn(2,4)
print(arr)
#random values that can be negative as well
```

arange → step-based values

linspace → fixed number of values

eye → identity matrix

rand → random floats (0–1)

randint → random integers in a range

randn → random values from bell curve distribution
